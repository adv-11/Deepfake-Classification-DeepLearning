```python
import torch

# Change this path for each model
path = r"C:\Users\Advait Shinde\Deepfake-Classification-DeepLearning\Gradio_App\final_dual_maxvit.pt"

ckpt = torch.load(path, map_location="cpu")

print("=== TYPE ===")
print(type(ckpt))

if isinstance(ckpt, dict):
    print("\n=== DICT KEYS ===")
    print(list(ckpt.keys()))
    
    # If it has a state_dict key
    if "state_dict" in ckpt:
        sd = ckpt["state_dict"]
    elif "model" in ckpt:
        sd = ckpt["model"]
    else:
        sd = ckpt  # the dict IS the state dict
    
    print("\n=== FIRST 5 LAYER NAMES ===")
    keys = list(sd.keys())
    for k in keys[:5]:
        print(f"  {k}: {sd[k].shape}")
    
    print("\n=== LAST 3 LAYER NAMES ===")
    for k in keys[-3:]:
        print(f"  {k}: {sd[k].shape}")

else:
    # It's a full saved model (torch.save(model, ...))
    print("\n=== FULL MODEL — ARCHITECTURE ===")
    print(ckpt)
    print("\n=== LAST LAYER ===")
    children = list(ckpt.children())
    print(children[-1] if children else "No children")
```
